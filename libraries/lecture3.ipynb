# 1. Привести различные способы создания объектов типа Series

```python
import pandas as pd
import numpy as np
```

## Для создания Series можно использовать:

### - списки Python

```python
data = pd.Series([1, 2, 3, 4, 5])
print(data)
```

### - массивы NumPy

```python
data = pd.Series(np.array([1, 2, 3, 4, 5]))
print(data)
```

### - скалярные значения

```python
data = pd.Series(12)
print(data)
```

### - словари

```python
data = pd.Series({'anna': 12, 'larry': 24, 'alex': 99, 'ronald': 44})
print(data)
```

## 2. Привести различные способы создания объектов типа DataFrame

## Для создания DataFrame можно использовать:

### - Объекты Series

```python
age_dict = {'anna': 12, 'larry': 24, 'alex': 99, 'ronald': 44}
height_dict = {'anna': 170, 'larry': 190, 'alex': 158, 'ronald': 180}
age = pd.Series(age_dict)
height = pd.Series(height_dict)
data = pd.DataFrame([age, height], index=['age', 'height'])
print(data)
```

### - Списки словарей

```python
data = pd.DataFrame([age_dict, height_dict], index=['age', 'height'])
print(data)
```

### - Словари объектов Series

```python
data = pd.DataFrame({
    'age': age,
    'height': height
})
print(data)
```

### - Двумерный массив NumPy

```python
data = pd.DataFrame(np.array([[12, 24, 99, 44], [170, 190, 158, 180]]), index=['age', 'height'], columns=['anna', 'larry', 'alex', 'ronald'])
print(data)
```

### - Структурированный массив NumPy

```python
people = np.array([
    ('anna', 12, 170),
    ('larry', 24, 190),
    ('alex', 99, 158),
    ('ronald', 44, 180)
], dtype=[('name', 'U30'), ('age', 'i4'), ('height', 'i4')])

data = pd.DataFrame(people)
print(data)
```

# 3. Объедините два объекта Series с неодинаковыми множествами ключей (индексов) так, чтобы вместо NaN было установлено значение 1

```python
height_dict = {'maria': 169, 'margareth': 171, 'anna': 170, 'larry': 190}

age = pd.Series(age_dict)
height = pd.Series(height_dict)
dataFrame = pd.DataFrame({
    'age': age,
    'height': height
}).fillna(1)

print(dataFrame)
```

# 4. Переписать пример с транслирование для DataFrame так, чтобы вычитание происходило по СТОЛБЦАМ

```python
from numpy.random import default_rng as rng
rows = 3
columns = 4 
df = pd.DataFrame(rng().integers(0, 10, (rows, columns)), columns=['a', 'b', 'c', 'd'])
print(df)

print(df.sub(df.iloc[0:rows, 0], axis=0))

print(df.sub(df.iloc[::2, 0], axis=0))
```

# 5. На примере объектов DataFrame продемонстрируйте использование методов ffill() и bfill()

```python
height_dict = {'maria': 169, 'margareth': 171, 'anna': 170, 'larry': 190}

age = pd.Series(age_dict)
height = pd.Series(height_dict)
dataFrame = pd.DataFrame({
    'age': age,
    'height': height
})

print("Исходный DataFrame:")
print(dataFrame)
print("DataFrame после ffill()")
print(dataFrame.ffill())
print("DataFrame после bfill()")
print(dataFrame.bfill())
```
