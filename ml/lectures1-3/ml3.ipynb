# Машинное обучение

ML решает следующую задачу - требуется подогнать заданный набор точек данными под соответсвующую функцию (отображение входа на выход), которая улавливает важные сигналы в данных и игнорирует помехи, а затем убедиться, что на новых данных функция работает хорошо

ML можно разделить на два типа:
- обучение с учителем (supervised learning)
- обучение без учителя (unsupervised learning)

### Обучение с учителем

Моделирует отношение между признаком и меткой. Служат для предсказания меток на основе маркированных обучающих данных. После построения модели можно использовать её для присвоения меток новым ранее неизвестным данным.

Обучение с учителем можно разделить на два типа:
- классификация (дискретная метка)
- регрессия (непрерывная метка)

### Обучение без учителя

Моделирование признака без меток. Такие модели служат для выявления структуры немаркированных данных.

Обучение без учителя можно разделить на два типа:
- кластеризация (выделяет отдельные группы данных)
- понижение размерности (поиск более сжатого представления данных)

### Частичное обучение (semi-supervised learning)

Характеризуется тем, что метки неполные. Не весь набор данных промаркирован.

### Обучение с подкреплением (reinforcement learning)

Система обучения улучшает свои характеристики на основе взаимодействия со средой. При этом взаимодействии система получает сигналы (функции наград), которые несут в себе информацию, насколько хорошо/плохо система решила задачу с точки зрения среды в которой она работает. Итоговая награда не станет максимальной.

### Процесс построения системы машинного обучения

1) ##### Предварительная обработка:
   - На вход поступают необработанные данные и метки
   - Происходит выбор признаков, масштабирование признаков
   - Понижение размерности
   - Выборка образцов
   - Деление на обучающий и тестовый набор
2) ##### Обучение:
   - Выбор модели
   - Кросс-валидация
   - Метрики эффективности
   - Оптимизация гиперпараметров.
3) ##### Оценка и формирование финальной модели
4) ##### Прогнозирование (использование)

### SciKit-learn

1) ##### Выбираем класс модели
2) ##### Устанавливаем гиперпараметры модели
3) ##### Создаем матрицу признаков и целевой массив
4) ##### Обучение модели .fit()
5) ##### Применить модель к новым данным:
   - .predict() - с учителем
   - .transform() - без учителя

## Обучение с учителем. Линейная регрессия.

### Простая линейная регрессия

y = ax + b, где y - таргет.

```python
import matplotlib.pyplot as plt
import numpy as np
```

```python
np.random.seed(1)
x = 10 * np.random.rand(50)
y = 2 * x - 1 + np.random.randn(50)

X = x[:, np.newaxis]

plt.scatter(x, y)
plt.show()
```

```python
from sklearn.linear_model import LinearRegression
model = LinearRegression(fit_intercept=False)

model.fit(X, y)

x_all = np.linspace(0, 10, 100)
plt.scatter(x, y)
plt.plot(x_all, model.coef_[0] * x_all + model.intercept_, 'r')
plt.show()

print(model.coef_[0])
print(model.intercept_)
```

```python
xfit = np.linspace(0, 10, 5)
yfit = model.predict(xfit[:, np.newaxis])
plt.scatter(x, y)
plt.scatter(xfit, yfit)
plt.plot(x_all, model.coef_[0] * x_all + model.intercept_, 'r')
plt.show()
```
